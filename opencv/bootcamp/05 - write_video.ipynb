```python
# import the library
import os
import cv2 as cv
import matplotlib.pyplot as plt

from zipfile import ZipFile
from urllib.request import urlretrieve

from IPython.display import YouTubeVideo, display, HTML
from base64 import b64encode

%matplotlib inline
```

```python
def download_and_unzip(url, save_path):
    print(f"Downloading and extracting assests....", end="")

    # Downloading zip file using urllib package.
    urlretrieve(url, save_path)

    try:
        # Extracting zip file using the zipfile package.
        with ZipFile(save_path) as z:
            # Extract ZIP file contents in the same directory.
            z.extractall(os.path.split(save_path)[0])

        print("Done")

    except Exception as e:
        print("\nInvalid file.", e)
```

```python
URL = r"https://www.dropbox.com/s/p8h7ckeo2dn1jtz/opencv_bootcamp_assets_NB6.zip?dl=1"

asset_zip_path = os.path.join(os.getcwd() + "/assets", f"opencv_bootcamp_assets_NB6.zip")

# Download if assest ZIP does not exists. 
if not os.path.exists(asset_zip_path):
    download_and_unzip(URL, asset_zip_path)  
```

```python
source = "assets/race_car.mp4"
cap = cv.VideoCapture(source)
if not cap.isOpened():
    print("Error opening video")
```

```python
_, frame = cap.read()
plt.imshow(cv.cvtColor(frame, cv.COLOR_BGR2RGB))
```

```python
width = int(cap.get(cv.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv.CAP_PROP_FPS))

avi = cv.VideoWriter("assets/race_car.avi", cv.VideoWriter_fourcc(*"MJPG"), fps, (width,height))
mp4 = cv.VideoWriter("assets/race_car.mp4", cv.VideoWriter_fourcc(*"XVID"), fps, (width,height))
```
